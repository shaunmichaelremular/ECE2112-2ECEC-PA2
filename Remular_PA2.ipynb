## A. REPRODUCIBLE NORMALIZATION PROBLEM

```python
import numpy as np
```

```python
np.random.seed(2112)
X = np.random.randint(10, 101, size=(5, 5))
X
```

```python
X
```

```python
X_normalized = (X-X.mean())/X.std()
X_normalized
```

```python
X_normalized.mean()
```

```python
round(X_normalized.std())
```

```python
print("Array: \n", X, "\n \n")
print("Normalized Array: \n", X_normalized, "\n \n")
print("Mean of Normalized Array: \n", round(X_normalized.mean()), "\n \n")
print("Standard Deviation of Normalized Array: \n", round(X_normalized.std()), "\n \n")
```

```python
np.save("X_normalized", X_normalized)
```

## B. CUBES DIVISIBLE BY 4 PROBLEM

```python
C = (np.arange(1,101,1)**3).reshape(10,10)
C
```

```python
div_by_4 = C[C%4==0]
div_by_4
```

```python
np.save("div_by_4", div_by_4)
```

## C. ABOVE-MEAN SQUARES PROBLEM

```python
S = (np.arange(1,37,1)**2)
S
```

```python
S_mean = S.mean()
S_mean
```

```python
above_mean = S[S>S_mean]
above_mean
```

```python
np.save("above_mean", above_mean)
```
